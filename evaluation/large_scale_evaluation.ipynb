# ReciCall Large-Scale Synthetic Evaluation

This notebook runs the synthetic experiment for balanced, unequal, cold-start, attendance, and tie-heavy classrooms. It compares ReciCall's priority queue with round-robin and random eligible selection.

```python
from pathlib import Path
import csv
import subprocess
import sys

ROOT = Path.cwd()
if not (ROOT / 'scripts').exists():
    ROOT = ROOT.parent

# Change --repetitions to 100 for the thesis experiment after reviewing the pilot run.
subprocess.run([
    sys.executable,
    str(ROOT / 'scripts' / 'run-large-scale-evaluation.py'),
    '--repetitions', '1',
], cwd=ROOT, check=True)
print('Experiment completed')
```

```python
results_path = ROOT / 'evaluation' / 'results' / 'large_scale_results.csv'
with results_path.open(newline='', encoding='utf-8') as handle:
    rows = list(csv.DictReader(handle))

print(f'Experiment rows: {len(rows)}')
print('Metrics: fairness, starvation prevention, eligibility compliance, coverage, distribution range, longest wait, formula validity, and repeat rate.')
for row in rows:
    if row['students'] == '1000' and row['algorithm'] == 'priority_queue':
        print(row)
```

## Compare algorithms

Priority compliance is expected to be highest for ReciCall because it is the oracle used by the simulation. Fairness and starvation prevention are the main behavior comparisons against the baselines.

```python
for scenario in sorted({row['scenario'] for row in rows}):
    print(f'\n{scenario}')
    for row in rows:
        if row['scenario'] == scenario and row['students'] == '1000':
            print(
                f"  {row['algorithm']}: fairness={row['avg_fairness_score']}, "
                f"starvation={row['avg_starvation_prevention_percent']}%, "
                f"coverage={row['avg_participation_coverage_percent']}%, "
                f"range={row['avg_distribution_range']}"
            )
```

## Thesis interpretation

Use the repeated-run averages and standard deviations in `large_scale_results.csv`. Synthetic testing evaluates algorithm behavior under controlled conditions; it does not replace classroom user acceptance testing or real teacher-reviewed accuracy labels.
